```python
import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx
import seaborn as sns
from mlxtend.frequent_patterns import apriori, association_rules

# Set seaborn style for better visuals
sns.set_theme(style="whitegrid")

# 1. Create the dataset (4 transactions)
dataset = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter', 'Jam'],
    ['Milk', 'Bread', 'Jam'],
    ['Milk', 'Butter']
]

# 2. One-hot encode with boolean values
unique_items = sorted(list(set(item for transaction in dataset for item in transaction)))
encoded_data = [{item: (True if item in transaction else False) for item in unique_items} for transaction in dataset]
df = pd.DataFrame(encoded_data)

# 3. Compute Frequent Itemsets
frequent_itemsets = apriori(df, min_support=0.5, use_colnames=True)
frequent_itemsets['length'] = frequent_itemsets['itemsets'].apply(lambda x: len(x))

# Explicitly sorting and saving to frequent_itemsets_sorted as requested
frequent_itemsets_sorted = frequent_itemsets.sort_values(by='support', ascending=False)

# 4. Generate Association Rules
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.5, num_itemsets=len(frequent_itemsets))

# --- Advanced Visualizations ---
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# Plot 1: Frequent Itemsets Support
itemset_labels = [', '.join(list(x)) for x in frequent_itemsets_sorted['itemsets']]
sns.barplot(
    x=frequent_itemsets_sorted['support'],
    y=itemset_labels,
    ax=axes[0],
    palette="Blues_r",
    hue=itemset_labels,
    legend=False
)
axes[0].set_xlabel('Support (Frequency of Occurrence)', fontsize=12)
axes[0].set_ylabel('Itemsets', fontsize=12)
axes[0].set_title('Top Frequent Itemsets Support', fontsize=14, fontweight='bold')
axes[0].grid(axis='x', linestyle='--', alpha=0.5)

# Plot 2: Directed Network Graph of Association Rules
if not rules.empty:
    G = nx.DiGraph()
    for idx, row in rules.iterrows():
        ant = ', '.join(list(row['antecedents']))
        con = ', '.join(list(row['consequents']))
        G.add_edge(ant, con, weight=row['lift'])

    pos = nx.circular_layout(G)

    # Draw nodes
    nx.draw_networkx_nodes(G, pos, ax=axes[1], node_size=1800, node_color='skyblue', edgecolors='navy', linewidths=1.5)
    nx.draw_networkx_labels(G, pos, ax=axes[1], font_size=11, font_weight='bold', font_family='sans-serif')

    # Draw edges with clear arrowheads indicating implication
    nx.draw_networkx_edges(
        G, pos, ax=axes[1],
        arrowstyle='-|>',
        arrowsize=25,
        edge_color='dimgray',
        width=2,
        connectionstyle="arc3,rad=0.15"
    )

    axes[1].set_title('Association Rules Network\n(Antecedent -> Consequent)', fontsize=14, fontweight='bold')
    axes[1].axis('off')
else:
    axes[1].text(0.5, 0.5, 'No rules met the threshold', ha='center', va='center', fontsize=12)
    axes[1].axis('off')

plt.tight_layout()
plt.show()

# Display Clean DataFrames
print("\n--- Sorted Frequent Itemsets ---")
display(frequent_itemsets_sorted)

print("\n--- Generated Rules (Sorted by Confidence) ---")
display(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].sort_values(by='confidence', ascending=False))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx
import seaborn as sns
from mlxtend.frequent_patterns import apriori, association_rules

# Set professional plot style
sns.set_theme(style="whitegrid")

# 1. Dataset of 4 transactions
dataset = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter', 'Jam'],
    ['Milk', 'Bread', 'Jam'],
    ['Milk', 'Butter']
]

# 2. Convert transactions into a boolean DataFrame
unique_items = sorted(list(set(item for transaction in dataset for item in transaction)))
encoded_data = [{item: (True if item in transaction else False) for item in unique_items} for transaction in dataset]
df = pd.DataFrame(encoded_data)

# 3. Find frequent itemsets using Apriori
frequent_itemsets = apriori(df, min_support=0.5, use_colnames=True)

# Sort frequent itemsets by support descending
frequent_itemsets_sorted = frequent_itemsets.sort_values(by='support', ascending=False)

# 4. Generate association rules
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.5, num_itemsets=len(frequent_itemsets))

# --- Visualizations ---
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# Plot 1: Horizontal Bar Chart of Frequent Itemsets
itemset_labels = [', '.join(list(x)) for x in frequent_itemsets_sorted['itemsets']]
sns.barplot(
    x=frequent_itemsets_sorted['support'],
    y=itemset_labels,
    ax=axes[0],
    palette="viridis",
    hue=itemset_labels,
    legend=False
)
axes[0].set_xlabel('Support', fontsize=12)
axes[0].set_title('Top Frequent Itemsets Support', fontsize=14, fontweight='bold')
axes[0].grid(axis='x', linestyle='--', alpha=0.5)

# Plot 2: Circular Directed Network Graph for Association Rules
if not rules.empty:
    G = nx.DiGraph()
    for idx, row in rules.iterrows():
        ant = ', '.join(list(row['antecedents']))
        con = ', '.join(list(row['consequents']))
        G.add_edge(ant, con, weight=row['lift'])

    pos = nx.circular_layout(G)

    nx.draw_networkx_nodes(G, pos, ax=axes[1], node_size=2000, node_color='lightgreen', edgecolors='darkgreen', linewidths=1.5)
    nx.draw_networkx_labels(G, pos, ax=axes[1], font_size=11, font_weight='bold')
    nx.draw_networkx_edges(
        G, pos, ax=axes[1],
        arrowstyle='-|>',
        arrowsize=25,
        edge_color='dimgray',
        width=2,
        connectionstyle="arc3,rad=0.15"
    )

    axes[1].set_title('Association Rules Network\n(Antecedent -> Consequent)', fontsize=14, fontweight='bold')
    axes[1].axis('off')
else:
    axes[1].text(0.5, 0.5, 'No rules met the threshold', ha='center', va='center', fontsize=12)
    axes[1].axis('off')

plt.tight_layout()
plt.show()

# Display Clean DataFrames
print("\n--- Sorted Frequent Itemsets ---")
display(frequent_itemsets_sorted)

print("\n--- Generated Rules (Sorted by Confidence) ---")
display(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].sort_values(by='confidence', ascending=False))
```
